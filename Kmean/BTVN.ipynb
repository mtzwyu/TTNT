```python
#BÀI 4: K-MEANS & K-NN DEMO
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import make_blobs, make_classification

print("--- DEMO 1: K-MEANS PHÂN CỤM KHÁCH HÀNG BÁN LẺ ---")
# Tạo dữ liệu giả lập: 300 khách hàng, 3 nhóm hành vi (Thu nhập & Điểm chi tiêu)
X_retail, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.5, random_state=42)

# Khởi tạo và huấn luyện mô hình K-Means với k=3
kmeans_model = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans_model.fit_predict(X_retail)
centers = kmeans_model.cluster_centers_

print(f"Tâm của 3 nhóm khách hàng (Thu nhập, Điểm chi tiêu):\n{centers}\n")

# ==========================================
print("--- DEMO 2: K-NN PHÂN LOẠI KHÁCH MUA TOUR DU LỊCH ---")
# Tạo dữ liệu giả lập: Độ tuổi, Thu nhập -> Mua tour cao cấp (1) hoặc không (0)
X_tour, y_tour = make_classification(n_samples=200, n_features=2, n_redundant=0,
                                     n_informative=2, n_clusters_per_class=1, random_state=42)

# Chia tập train/test
X_train, X_test, y_train, y_test = train_test_split(X_tour, y_tour, test_size=0.25, random_state=42)

# Khởi tạo và huấn luyện mô hình K-NN với k=5
knn_model = KNeighborsClassifier(n_neighbors=5, weights='uniform')
knn_model.fit(X_train, y_train)

# Dự đoán và đánh giá
y_pred = knn_model.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"Độ chính xác của hệ thống dự đoán khách hàng mua Tour: {acc * 100:.2f}%")

# Dự đoán một khách hàng mới cụ thể
new_customer = np.array([[1.5, -2.0]]) # Giá trị giả định đã scale
pred_new = knn_model.predict(new_customer)
print(f"Dự đoán cho khách hàng mới {new_customer[0]}: {'Sẽ mua' if pred_new[0] == 1 else 'Không mua'} tour cao cấp.")
```

```python
# BÀI 5: THUẬT TOÁN LÝ THUYẾT DEMO
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, accuracy_score

print("--- DEMO 1: LINEAR REGRESSION DỰ ĐOÁN DOANH THU ---")
# Tạo dữ liệu: X = Chi phí quảng cáo (triệu VNĐ), y = Doanh thu (triệu VNĐ)
np.random.seed(42)
X_budget = 10 * np.random.rand(100, 1) # Ngân sách từ 0 -> 10 triệu
y_revenue = 5 + 3 * X_budget + np.random.randn(100, 1) * 2 # Doanh thu có nhiễu

# Chia tập dữ liệu
X_train_lr, X_test_lr, y_train_lr, y_test_lr = train_test_split(X_budget, y_revenue, test_size=0.2, random_state=42)

# Huấn luyện mô hình
lr_model = LinearRegression()
lr_model.fit(X_train_lr, y_train_lr)

# Dự đoán
y_pred_lr = lr_model.predict(X_test_lr)
mse = mean_squared_error(y_test_lr, y_pred_lr)
print(f"Hệ số góc (Trọng số w): {lr_model.coef_[0][0]:.2f}")
print(f"Sai số toàn phương trung bình (MSE): {mse:.2f}")
print(f"Dự đoán doanh thu nếu đầu tư 7 triệu quảng cáo: {lr_model.predict([[7]])[0][0]:.2f} triệu VNĐ\n")

# ==========================================
print("--- DEMO 2: DECISION TREE ĐÁNH GIÁ DỊCH VỤ ---")
# Tạo dữ liệu: Đặc trưng 1 (Thời gian chờ), Đặc trưng 2 (Điểm phục vụ)
X_service, y_satisfaction = make_classification(n_samples=150, n_features=2, n_redundant=0,
                                                n_informative=2, random_state=42)

# Chia tập dữ liệu
X_train_dt, X_test_dt, y_train_dt, y_test_dt = train_test_split(X_service, y_satisfaction, test_size=0.3, random_state=42)

# Huấn luyện mô hình Cây quyết định
dt_model = DecisionTreeClassifier(max_depth=3, random_state=42)
dt_model.fit(X_train_dt, y_train_dt)

# Đánh giá mô hình
y_pred_dt = dt_model.predict(X_test_dt)
acc_dt = accuracy_score(y_test_dt, y_pred_dt)
print(f"Độ chính xác của mô hình Cây quyết định: {acc_dt * 100:.2f}%")
print(f"Độ quan trọng của các đặc trưng (Feature Importances): {dt_model.feature_importances_}")
```
